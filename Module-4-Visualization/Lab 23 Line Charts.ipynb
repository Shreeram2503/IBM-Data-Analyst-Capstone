### Task 1: Trends in Compensation Over Age Groups


##### 1. Line Chart of Median `ConvertedCompYearly` by Age Group


- Track how the median yearly compensation (ConvertedCompYearly) changes across different age groups.

- Use a line chart to visualize these trends.


```python
## Write your code here
##Write your code here
def convert_age(age):
    if pd.isna(age):
        return None

    age = str(age).lower().replace('years old', '').strip()

    if 'under' in age:
        return 17

    elif '-' in age:
        parts = age.split('-')
        return (int(parts[0]) + int(parts[1])) / 2
    elif age.isdigit():
        return int(age)

    return None



df_line = df[
    [
        'Age',
        'ConvertedCompYearly'
    ]
].copy()

df_line['Age_convert'] = df_line[
    'Age'
].apply(convert_age)

df_line['ConvertedCompYearly'] = pd.to_numeric(
    df_line['ConvertedCompYearly'],
    errors='coerce'
)

df_line = df_line.dropna()

# Optional outlier filter
df_line = df_line[
    df_line['ConvertedCompYearly'] <= 100000
]

bins = [15,25,35,45,55,65]

labels = [
    '15-25',
    '26-35',
    '36-45',
    '46-55',
    '56-65'
]

df_line['AgeGroup'] = pd.cut(
    df_line['Age_convert'],
    bins=bins,
    labels=labels
)

grouped = df_line.groupby(
    'AgeGroup'
)[
    'ConvertedCompYearly'
].median()

grouped.plot(
    kind='line',
    marker='o',
    figsize=(10,6)
)

plt.title(
    "Median Compensation by Age Group"
)

plt.xlabel("Age Group")

plt.ylabel("Median Compensation")

plt.grid(True)

plt.show()
```

##### 2. Line Chart of Median `ConvertedCompYearly` for Ages 25 to 45


For a closer look, plot a line chart focusing on the median compensation for respondents between ages 25 and 45.


```python
##Write your code here

df_age = df[
    [
        'Age',
        'ConvertedCompYearly'
    ]
].copy()

df_age['Age_convert'] = df_age[
    'Age'
].apply(convert_age)

df_age['ConvertedCompYearly'] = pd.to_numeric(
    df_age['ConvertedCompYearly'],
    errors='coerce'
)

df_age = df_age.dropna()

df_age = df_age[
    df_age['ConvertedCompYearly'] <= 100000
]

age_filtered = df_age[
    (df_age['Age_convert'] >= 25) &
    (df_age['Age_convert'] <= 45)
]

grouped = age_filtered.groupby(
    'Age_convert'
)[
    'ConvertedCompYearly'
].median()

grouped.plot(
    kind='line',
    marker='o',
    figsize=(10,6)
)

plt.title(
    "Median Compensation for Ages 25 to 45"
)

plt.xlabel("Age")

plt.ylabel("Median Compensation")

plt.grid(True)

plt.show()
```

### Task 2: Trends in Job Satisfaction by Experience Level



##### 1. Line Chart of Job Satisfaction (`JobSatPoints_6`) by Experience Level



- Use a column that approximates experience level to analyze how job satisfaction changes with experience.

- If needed, substitute an available experience-related column for `Experience`.


```python
## Write your code here
##Write your code here

df_exp = df[
    [
        'WorkExp',
        'JobSatPoints_6'
    ]
].copy()

df_exp['WorkExp'] = pd.to_numeric(
    df_exp['WorkExp'],
    errors='coerce'
)

df_exp['JobSatPoints_6'] = pd.to_numeric(
    df_exp['JobSatPoints_6'],
    errors='coerce'
)

df_exp = df_exp.dropna()

grouped = df_exp.groupby(
    'WorkExp'
)[
    'JobSatPoints_6'
].median()

grouped.plot(
    kind='line',
    marker='o',
    figsize=(10,6)
)

plt.title(
    "Job Satisfaction by Experience Level"
)

plt.xlabel("Years of Experience")

plt.ylabel("Median Job Satisfaction")

plt.grid(True)

plt.show()
```

### Task 3: Trends in Job Satisfaction and Compensation by Experience


##### 1.Line Chart of Median ConvertedCompYearly Over Experience Level

- This line chart will track how median compensation (`ConvertedCompYearly`) changes with increasing experience.

- Use a column such as `WorkExp` or another relevant experience-related column.


```python
## Write your code here
df_comp_exp = df[['WorkExp', 'ConvertedCompYearly']].copy()

df_comp_exp['ConvertedCompYearly'] = pd.to_numeric(df_comp_exp['ConvertedCompYearly'], errors='coerce')

df_comp_exp['WorkExp'] = pd.to_numeric(df_comp_exp['WorkExp'], errors='coerce')

df_comp_exp = df_comp_exp.dropna()

df_comp_exp = df_comp_exp[df_comp_exp['ConvertedCompYearly'] <= 1000000]

grouped = df_comp_exp.groupby('WorkExp')['ConvertedCompYearly'].median()

grouped.plot(
    kind='line',
    marker='o',
    figsize=(10,6)
)

plt.title(
    "Median Compensation by Experience Level"
)

plt.xlabel("Years of Experience")

plt.ylabel("Median Compensation")

plt.grid(True)

plt.show()
```

##### 2.Line Chart of Job Satisfaction (`JobSatPoints_6`) Across Experience Levels

- Create a line chart to explore trends in job satisfaction (`JobSatPoints_6`) based on experience level.

- This chart will provide insight into how satisfaction correlates with experience over time


```python
##Write your code here

df_jobsat = df[
    [
        'WorkExp',
        'JobSatPoints_6'
    ]
].copy()

df_jobsat['WorkExp'] = pd.to_numeric(
    df_jobsat['WorkExp'],
    errors='coerce'
)

df_jobsat['JobSatPoints_6'] = pd.to_numeric(
    df_jobsat['JobSatPoints_6'],
    errors='coerce'
)

df_jobsat = df_jobsat.dropna()

grouped = df_jobsat.groupby(
    'WorkExp'
)[
    'JobSatPoints_6'
].median()

grouped.plot(
    kind='line',
    marker='o',
    figsize=(10,6)
)

plt.title(
    "Job Satisfaction Across Experience Levels"
)

plt.xlabel("Years of Experience")

plt.ylabel("Median Job Satisfaction")

plt.grid(True)

plt.show()
```
